```python
# WAP in Python using all list operations

def list_operations():
    # 1. Create a list
    my_list = [10, 20, 30, 40, 50]
    print("Original List:", my_list)

    # 2. Indexing
    print("First element:", my_list[0])
    print("Last element:", my_list[-1])

    # 3. Slicing
    print("Slice [1:4]:", my_list[1:4])

    # 4. Append
    my_list.append(60)
    print("After append:", my_list)

    # 5. Insert
    my_list.insert(2, 25)
    print("After insert at index 2:", my_list)

    # 6. Extend
    my_list.extend([70, 80])
    print("After extend:", my_list)

    # 7. Update
    my_list[0] = 5
    print("After update first element:", my_list)

    # 8. Remove
    my_list.remove(25)
    print("After remove 25:", my_list)

    # 9. Pop
    popped = my_list.pop()
    print("After pop:", my_list, "| Popped element:", popped)

    # 10. Search (index)
    print("Index of 40:", my_list.index(40))

    # 11. Count
    print("Count of 20:", my_list.count(20))

    # 12. Sort
    my_list.sort()
    print("Sorted list:", my_list)

    # 13. Reverse
    my_list.reverse()
    print("Reversed list:", my_list)

    # 14. Copy
    copied_list = my_list.copy()
    print("Copied list:", copied_list)

    # 15. Clear
    my_list.clear()
    print("After clear:", my_list)

if __name__ == "__main__":
    list_operations()
```

```python
def lowest_temp(daily_temp):
     lowest_temp = daily_temp[0]
     for temp in daily_temp:
         if temp < lowest_temp:
             lowest_temp = daily_temp
             return lowest_temp
weekly_temp=[45,28,33]
print("coldest temp of week",lowest_temp(weekly_temp))
```

```python
def duplicate_item(cart_item):
      n = len(cart_item)
      for i in range(n):
          for j in range(i+1,n):
              if cart_item == cart_item[j]:
                  return cart_item[i]
      return "No duplicate"
cart = ["milk","coffee","coffee","samosa","bread","bread"]
print("duplicate item",duplicate_item(cart))
```

```python
import tkinter as tk
import time

def update_time():
    current_time=time.strftime('%I:%M:%S %p')
    clock_label.config(text=current_time)
    clock_label.after(1000,update_time)

root =tk.Tk()
root.title("Digital Clock")
root.geometry("400x150")
root.configure(bg="Black")
root.resizable(False,False)

clock_label=tk.Label(
    root,
    font=("Arial",36,"bold"),
    bg="black",
    fg="#00FF00"
)
clock_label.pack(expand=True)
update_time()
root.mainloop()
```
